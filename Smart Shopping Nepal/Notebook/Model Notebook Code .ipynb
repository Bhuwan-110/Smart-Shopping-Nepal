```python
# Saving amount ra saving percentage pani automatic calculate garne. Summary table ma saving add garne

summary = []  # sabai product ko comparison result rakhne list banaune

for key in test_df["product_key"].unique():  # harek unique product ko lagi loop chalaune
    options = test_df[test_df["product_key"] == key].copy()  # tyo product ko sabai store listings chhantne
    options = options.sort_values("total_price_npr")  # sasto dekhi mahango order ma milाउने

    cheapest = options.iloc[0]  # sabai bhanda sasto listing chhantne
    highest_price = options["total_price_npr"].max()  # sabai store madhye highest price nikalne
    saving = highest_price - cheapest["total_price_npr"]  # highest ra cheapest price ko difference nikalne
    saving_percent = (saving / highest_price) * 100 if highest_price > 0 else 0  # saving percentage calculate garne

    summary.append({  # result lai summary list ma add garne
        "product_name": cheapest["product_name"],  # product ko naam rakhne
        "cheapest_store": cheapest["platform"],  # cheapest store ko naam rakhne
        "total_price_npr": cheapest["total_price_npr"],  # cheapest total price rakhne
        "number_of_stores": len(options),  # product kati store ma chha count garne
        "saving_npr": saving,  # kati rupaiya bachchha rakhne
        "saving_percent": round(saving_percent, 2)  # saving percentage rakhne
    })

summary_df = pd.DataFrame(summary)  # summary list lai Pandas DataFrame banaune
display(summary_df)  # updated comparison table dekhaune
```
