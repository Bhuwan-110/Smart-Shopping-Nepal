```python
import pandas as pd
import numpy as np
import sklearn

print("SmartShopping Nepal")
print("Project setup successful!")
print("Ready to start data collection.")
```

```python
products = pd.DataFrame([
    {
        "product_name": "Samsung Galaxy A56",
        "category": "Mobile Phone",
        "brand": "Samsung",
        "model": "Galaxy A56",
        "variant": "8GB/256GB"
    },
    {
        "product_name": "HP Laptop",
        "category": "Laptop",
        "brand": "HP",
        "model": "Pavilion",
        "variant": "16GB/512GB"
    },
    {
        "product_name": "Wireless Earbuds",
        "category": "Earbuds",
        "brand": "Example Brand",
        "model": "Example Model",
        "variant": "Standard"
    }
])
products
```

```python
products["platform"] = None
products["price_npr"] = None
products["shipping_cost_npr"] = None
products["rating"] = None
products["review_count"] = None
products["product_url"] = None

products
```

```python
print("Dataset shape:", products.shape)  # Dataset ma kati rows र columns xan dekhaunxa

print("\nColumn names:")  # Column names dekhauna heading print garxa
print(products.columns.tolist())  # sabai column ka name haru list ma dekhaunxa

print("\nMissing values:")  # Missing values ko result dekhauna heading print garxa
print(products.isnull().sum())  # each column मा kati values khali xan ganxa
```

```python
print("Duplicate rows:", products.duplicated().sum()) # same/duplicate ayaka rows ko no. dekhaunxa
print(products.duplicated()) # every row duplicate vaya true, na vaya false dekhaunxa.
```

```python
#Product Names safa garne (Text Cleaning) garne.

products["product_name"] = products["product_name"].str.strip()  # Product name ko agadi ra pachhadiko anawasyak space hatauchha

products["brand"] = products["brand"].str.strip()  # Brand ko agadi ra pachhadiko anawasyak space hatauchha

products["model"] = products["model"].str.strip()  # Model ko agadi ra pachhadiko anawasyak space hatauchha

print(products[["product_name", "brand", "model"]])  # Product name, brand ra model ko safaa gareko data dekhaunchha
```

```python
# Product Product name ma capital ra small letters farak-farak huna sakchhan. Hami matching sajilo 
# banauna euta naya column product_name_clean banaauchhau, jasle original name lai change gardaina.

products["product_name_clean"] = products["product_name"].str.lower()  # Product name lai sabai small letters ma badalera naya column banaauchha

print(products[["product_name", "product_name_clean"]])  # Original ra clean gareko product name compare garera dekhaunchha
```
